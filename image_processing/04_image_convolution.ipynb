```python
# image blurring manually
import numpy as np
import cv2

# Load the image (OpenCV loads in BGR format by default)
img_matrix_bgr = cv2.imread('myimage.png')

# (Optional) Convert from BGR to RGB for consistent representation
img_matrix = cv2.cvtColor(img_matrix_bgr, cv2.COLOR_BGR2RGB)

# Print the shape of the matrix and some pixel data
print(f"Image shape (BGR): {img_matrix_bgr.shape}")
print(f"Top-left pixel values (RGB): {img_matrix[0, 0]}")

# Apply a Gaussian blur to the image
gaussian_kernel_1d = cv2.getGaussianKernel(ksize=3, sigma=1)
gaussian_kernel = gaussian_kernel_1d @ gaussian_kernel_1d.T  # Create a 2D Gaussian kernel

# Manually apply the Gaussian blur using convolution
n,m,t= img_matrix.shape
blurred_img = np.zeros_like(img_matrix)
for i in range (1,n-2):
    for j in range (1,m-2):
        for c in range (3): 
            patch = img_matrix[i-1:i+2, j-1:j+2, c]
            blurred_img[i, j, c] = np.sum(patch * gaussian_kernel) # For each color channel
print ("Blurring completed.")
# Convert back to BGR format for saving
blurred_img_bgr = cv2.cvtColor(blurred_img, cv2.COLOR_RGB2BGR)
# Save the blurred image
cv2.imwrite('blurred_image_gaussian.png', blurred_img_bgr)
print("Blurred image saved as 'blurred_image_gaussian.png'.")
```

```python
# image blurring manually using moving avg kernel
import numpy as np
import cv2

k = 9
# Load the image (OpenCV loads in BGR format by default)
img_matrix_bgr = cv2.imread('myimage.png')

# (Optional) Convert from BGR to RGB for consistent representation
img_matrix = cv2.cvtColor(img_matrix_bgr, cv2.COLOR_BGR2RGB)

# Print the shape of the matrix and some pixel data
print(f"Image shape (BGR): {img_matrix_bgr.shape}")
print(f"Top-left pixel values (RGB): {img_matrix[0, 0]}")

# Apply a moving average blur to the image
kernel = np.ones((k,k))/float(k**2)

# Manually apply the Gaussian blur using convolution
n,m,t= img_matrix.shape
blurred_img = np.zeros_like(img_matrix)
for i in range (k//2,n-k//2):
    for j in range (k//2,m-k//2):
        for c in range (3): 
            patch = img_matrix[i-k//2:i+k//2 + 1, j-k//2:j+k//2 + 1, c]
            blurred_img[i, j, c] = np.sum(patch * kernel) # For each color channel
print ("Blurring completed.")
# Convert back to BGR format for saving
blurred_img_bgr = cv2.cvtColor(blurred_img, cv2.COLOR_RGB2BGR)
# Save the blurred image
cv2.imwrite('blurred_image_moving_avg.png', blurred_img_bgr)
print("Blurred image saved as 'blurred_image_moving_avg.png'.")
```
