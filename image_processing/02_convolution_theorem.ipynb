```python
# 02_convolution_theorem.ipynb
import numpy as np
import matplotlib.pyplot as plt

x = np.random.rand(1000)*100
h = np.ones(1000)
plt.plot(x, label='x')
plt.xlabel('Sample Index x')
plt.ylabel('Amplitude')
plt.show()
plt.plot(h, label='h')  
plt.xlabel('Sample Index h')
plt.ylabel('Amplitude')
plt.show()
y = np.convolve(x, h, mode='full')
plt.plot(y, label='y')
plt.show()
Y = np.fft.fft(y)
X = np.fft.fft(x, n=len(Y))
H = np.fft.fft(h, n=len(Y))
Y_ = X * H
plt.plot(Y, label='Y')
plt.xlabel('Frequency') 
plt.ylabel('Magnitude')
plt.show()
plt.plot(Y_, label='Y_')
plt.xlabel('Frequency') 
plt.ylabel('Magnitude')
plt.show()
if np.allclose(Y, Y_):
    print("The convolution theorem holds: Y is approximately equal to Y_.")
else:    print("The convolution theorem does not hold: Y is not approximately equal to Y_.")
```
