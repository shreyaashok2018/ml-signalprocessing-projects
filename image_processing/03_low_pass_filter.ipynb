```python
# low pass filter
# moving average filter
import numpy as np
import matplotlib.pyplot as plt
fs = 1000         # sampling frequency (Hz)
duration = 1       # seconds
f = 5             # sine frequency (Hz)

t = np.arange(0, duration, 1/fs)
x = np.sin(2*np.pi*f*t)

noise = 0.3 * np.random.randn(len(t))
x_noisy = x + noise
plt.plot(t[:], x_noisy[:])
plt.title("Noisy Sine Wave")
plt.show()

h = np.ones(15)/15
x_filtered = np.convolve(x_noisy, h, mode='same')
plt.plot(t[:], x_filtered[:])
plt.title("Filtered Sine Wave")
plt.show()

y = np.fft.fft(x_noisy)
freqs = np.fft.fftfreq(len(x_noisy), 1/fs)
plt.plot(freqs[:len(freqs)//2], np.abs(y[:len(y)//2]))
plt.title("Frequency Spectrum of Noisy Signal")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.show()

y_filtered = np.fft.fft(x_filtered)
plt.plot(freqs[:len(freqs)//2], np.abs(y_filtered[:len(y_filtered)//2]))
plt.title("Frequency Spectrum of Filtered Signal")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.show()
```

```python
# low pass filter
# gaussian filter
import numpy as np
import matplotlib.pyplot as plt

fs = 1000         # sampling frequency (Hz)
duration = 1       # seconds
f = 5             # sine frequency (Hz)

t = np.arange(0, duration, 1/fs)
x = np.sin(2*np.pi*f*t)

noise = 0.3 * np.random.randn(len(t))
x_noisy = x + noise
plt.plot(t[:], x_noisy[:])
plt.title("Noisy Sine Wave")
plt.show()

# gaussian kernel
v = np.array([-1,0,1])
h = np.exp((- v/1)**2)
h = h/h.sum()
x_filtered = np.convolve(x_noisy, h, mode='same')
plt.plot(t[:], x_filtered[:])
plt.title("Filtered Sine Wave")
plt.show()

y = np.fft.fft(x_noisy)
freqs = np.fft.fftfreq(len(x_noisy), 1/fs)
plt.plot(freqs[:len(freqs)//2], np.abs(y[:len(y)//2]))
plt.title("Frequency Spectrum of Noisy Signal")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.show()

y_filtered = np.fft.fft(x_filtered)
plt.plot(freqs[:len(freqs)//2], np.abs(y_filtered[:len(y_filtered)//2]))
plt.title("Frequency Spectrum of Filtered Signal")
plt.xlabel("Frequency (Hz)")
plt.ylabel("Magnitude")
plt.show()
```
